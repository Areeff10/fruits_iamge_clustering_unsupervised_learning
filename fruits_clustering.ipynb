```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data=np.load('fruits_300.npy')
data
```

```python
data.shape
```

```python
plt.imshow(data[0],cmap='gray_r')
plt.show()
```

```python
fig, axs=plt.subplots(1, 5, figsize=(15, 15))
axs[0].imshow(data[0], cmap='gray_r')
axs[1].imshow(data[1], cmap='gray_r')
axs[2].imshow(data[100], cmap='gray_r')
axs[3].imshow(data[200], cmap='gray_r')
axs[4].imshow(data[299], cmap='gray_r')
plt.show()
```

```python
data_2d = data.reshape(-1, 100*100)
data_2d.shape
```

```python
km = KMeans(init='random', random_state=2207)
km.fit(data_2d)
```

```python
inertia = []
for k in range(2, 7):
    km = KMeans(n_clusters=k, random_state=2207)
    km.fit(data_2d)
    inertia.append(km.inertia_)

plt.plot(range(2, 7), inertia)
plt.xlabel('k')
plt.ylabel('inertia')
plt.show()
```

```python
km = KMeans(init='random', n_clusters=3, random_state=2207)
km.fit(data_2d)
```

```python
print(km.labels_)
```

```python
import matplotlib.pyplot as plt

def draw_fruits(images, row=1, col=10):
    plt.figure(figsize=(col, row))
    for i in range(row * col):
        plt.subplot(row, col, i + 1)
        plt.imshow(images[i], cmap='gray')
        plt.axis("off")
    plt.show()
```

```python
draw_fruits(data[km.labels_==0])
```

```python
draw_fruits(data[km.labels_==1])
```

```python
draw_fruits(data[km.labels_==2])
```
